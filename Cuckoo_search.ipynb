```python
import numpy as np
import math

def fitness(x):
    return sum(x**2)

def levy_flight():
    beta = 1.5

    sigma = (
        (math.gamma(1 + beta) * math.sin(math.pi * beta / 2)) /
        (math.gamma((1 + beta) / 2) * beta *
         2**((beta - 1) / 2))
    ) ** (1 / beta)

    u = np.random.normal(0, sigma)
    v = np.random.normal(0, 1)

    step = u / (abs(v)**(1 / beta) + 1e-10)

    return 0.01 * step

n = 5
iterations = 20
dimension = 2

nests = np.random.uniform(-10, 10, (n, dimension))

scores = np.array([fitness(x) for x in nests])

best_index = np.argmin(scores)
best_solution = nests[best_index].copy()
best_fitness = scores[best_index]

print("Initial Best Fitness:", best_fitness)

for iteration in range(iterations):

    for i in range(n):

        new_solution = (
            nests[i] +
            levy_flight() * np.random.randn(dimension)
        )

        new_solution = np.clip(new_solution, -10, 10)

        new_fitness = fitness(new_solution)

        j = np.random.randint(n)

        if new_fitness < scores[j]:
            nests[j] = new_solution.copy()
            scores[j] = new_fitness

        best_index = np.argmin(scores)

        if scores[best_index] < best_fitness:
            best_fitness = scores[best_index]
            best_solution = nests[best_index].copy()

    print("Iteration:", iteration + 1,
          "Best Fitness:", round(best_fitness, 6))

print("\nBest Solution:", best_solution)
print("Best Fitness:", best_fitness)
```
