```python

import numpy as np
import random

n = 10

np.random.seed(42)
distance = np.random.randint(10, 100, size=(n, n))

np.fill_diagonal(distance, 0)

pheromone = np.ones((n, n))

alpha = 1
beta = 2

def calculate_probability(current, unvisited):

    numerator = []

    for city in unvisited:
        value = (pheromone[current][city] ** alpha) * \
                ((1 / distance[current][city]) ** beta)

        numerator.append(value)

    total = sum(numerator)

    probabilities = [value / total for value in numerator]

    return probabilities

def construct_route():

    start = random.randint(0, n - 1)

    route = [start]
    unvisited = list(range(n))
    unvisited.remove(start)

    print("Starting City:", start)

    while unvisited:

        current = route[-1]

        probabilities = calculate_probability(
            current, unvisited
        )

        print("\nCurrent City:", current)
        print("Unvisited Cities:", unvisited)
        print("Selection Probabilities:")

        for city, probability in zip(unvisited, probabilities):
            print(f"City {city}: {probability:.4f}")

        next_city = random.choices(
            unvisited,
            weights=probabilities,
            k=1
        )[0]

        route.append(next_city)
        unvisited.remove(next_city)

    return route


for i in range(3):

    print("\nRoute", i + 1)

    route = construct_route()

    print("Final Route:", route)
    print("Number of Cities:", len(route))
    print("Unique Cities:", len(set(route)))

    assert len(route) == n
    assert len(set(route)) == n
    assert set(route) == set(range(n))

    print("Valid Route: Yes")
```
